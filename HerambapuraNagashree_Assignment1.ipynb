Question(2):
Project Title: AI-Powered pet assistant that monitors pet health and behavior

```python
# Modified setup for AMD/CPU environment
import torch
from diffusers import StableDiffusionPipeline

model = "runwayml/stable-diffusion-v1-5"

dev = "cuda" if torch.cuda.is_available() else "cpu"
dt = torch.float16 if dev == "cuda" else torch.float32

pipe = StableDiffusionPipeline.from_pretrained(
    model,
    torch_dtype=dt,
    safety_checker=None,
    requires_safety_checker=False
)

pipe = pipe.to(dev)

print("Loaded on:", dev)
```

**Question(2) - Part (a) - Social Media Marketing Advertisement:**

```python
from transformers import pipeline, set_seed
set_seed(10)

gen = pipeline(
    "text-generation",
    model="gpt2"
)

prod = "AI-powered smart pet assistant that monitors pet health and behavior"

txt = f"""
A social media marketing advertisement for PetBuddy: {prod} .

Features:
- Monitors pet health and behavior
- Tracks sleep and daily activity
- Sends alerts to pet owners
- Helps owners understand pet wellness

Advertisement: PetBuddy is an AI-powered smart pet assistant that helps pet owners monitor
their pet's daily health and behavior.It tracks activity levels,sleep patterns,
and unusual changes while providing useful alerts.PetBuddy helps owners better
understand their pets and support their overall well-being.
"""

out = gen(
    txt,
    max_new_tokens=120,
    temperature=0.8,
    do_sample=True
)

ad = out[0]["generated_text"]

print(ad)
```

**Part b - Promotional Image Generation:**

```python
import torch
torch.manual_seed(0)

from IPython.display import display

pets =[
    "golden retriever with smart health collar in living room",
    "orange cat sitting near phone with pet activity app",
    "beagle dog wearing small AI pet tracker",
    "white rabbit near smart pet monitoring device",
    "puppy and kitten with mobile pet wellness dashboard"
]

for n in range(len(pets)):
    my_prompt = pets[n] + ",realistic photo,clean background,social media ad,high quality"

    my_img = pipe(
        my_prompt,
        num_inference_steps=20,
        guidance_scale=7.5
    ).images[0]

    img_name = "my_pet_ad_"+str(n + 1)+".png"

    my_img.save(img_name)

    print("image saved:",img_name)

    display(my_img)
```

To display all the generated images side by side

```python
from PIL import Image
import matplotlib.pyplot as plt

files = [
    "my_pet_ad_1.png",
    "my_pet_ad_2.png",
    "my_pet_ad_3.png",
    "my_pet_ad_4.png",
    "my_pet_ad_5.png"
]

names = [
    "Golden Retriever",
    "Persian Cat",
    "Beagle",
    "White Rabbit",
    "Puppy & Kitten"
]

plt.figure(figsize=(16,5))

for i,f in enumerate(files):

    pic = Image.open(f)

    plt.subplot(1, len(files), i + 1)
    plt.imshow(pic)
    plt.axis("off")
    plt.title(names[i])

plt.tight_layout()
plt.show()
```

**Part (c) - Generation of Audio - text to speech**

```python
from gtts import gTTS
from IPython.display import Audio

msg = """
PetBuddy is an AI-powered smart pet assistant that helps pet owners monitor
their pet's daily health and behavior.It tracks activity levels,sleep patterns,
and unusual changes while providing useful alerts.PetBuddy helps owners better
understand their pets and support their overall well-being.
"""

audio_file = gTTS(
    text=msg,
    lang="en"
)

audio_file.save("pet_voice.mp3")

print("Audio file generated successfully")

Audio("pet_voice.mp3")
```

Part(d) - Ethical Considerations Evaluations:



(a) Yes,AI-generated advertisements may overstate a product's capabilities or make false claims. 
An AI pet assistant,for example; can monitor behavior and activities,but it shouldn't be used in place of a veterinarian or a medical diagnostic. 
And also before being published,every advertisement should be examined.

(b) If the AI produces advertisements that exclusively highlight particular pet breeds,owner types,lifestyles,or living circumstances, bias may become apparent.This could be unfair representation and exclude certain user groups.Biases in the training data can also be detected by AI algorithms.

(c) Before publishing,users should ensure that AI-generated advertisements are accurate and do not contain any misleading promises.The content must be evaluated for bias,fairness,and inclusivity.Product descriptions,visuals,and advertising messaging should all be accurate representations of what the product can do.It must go through a final review and approval process,before the advertisement goes live.
