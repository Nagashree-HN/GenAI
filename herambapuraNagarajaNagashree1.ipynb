```python
# --- STEP 1: Environment Setup and Model Initialization ---
import torch
from transformers import pipeline
from diffusers import StableDiffusionPipeline
from gtts import gTTS
import os

# Setting configuration for CPU execution
# Using float32 for maximum compatibility across various local hardware environments
device = "cpu"
precision = torch.float32

print("Initializing AI components...")

# Load the text generation model (LLM) for ad copy creation
ad_engine = pipeline("text-generation", model="distilgpt2")

# Load the image generation model (Stable Diffusion)
# Note: 'segmind/small-sd' is selected for its lightweight architecture
model_id = "segmind/small-sd"
image_engine = StableDiffusionPipeline.from_pretrained(
    model_id, 
    torch_dtype=precision
)
image_engine = image_engine.to(device)

print("AI components initialized and ready.")

# --- STEP 2: Marketing Ad Generation (Text) ---
def create_marketing_copy(brand_focus):
    """Generates a unique advertisement hook for the AI powered pet assistant that monitors health and behavior."""
    seed_text = f"Introducing our new smart assistant for your pet, focusing on {brand_focus}."
    generated_content = ad_engine(seed_text, max_length=60, num_return_sequences=1)
    return generated_content[0]['generated_text']

marketing_text = create_marketing_copy("safety and health tracking")
print(f"\n--- Generated Advertisement --- \n{marketing_text}")

# --- STEP 3: Product Visual Generation (Images) ---
def generate_product_visuals(prompt_description, num_variations=4):
    """Creates visual representations of the product using latent diffusion."""
    print(f"\nGenerating {num_variations} visual concepts for: {prompt_description}")
    visuals = image_engine(prompt_description, num_images_per_prompt=num_variations).images
    
    for idx, visual in enumerate(visuals):
        file_path = f"pet_assistant_design_{idx+1}.png"
        visual.save(file_path)
        print(f"Visual concept saved: {file_path}")

generate_product_visuals("High-tech smart pet assistant device, modern minimalist design")

# --- STEP 4: Audio Narration Production ---
def produce_audio_narration(text_content, output_filename="product_pitch.mp3"):
    """Converts the marketing text into a synthesized audio narration."""
    audio_narrator = gTTS(text=text_content, lang='en')
    audio_narrator.save(output_filename)
    print(f"\nNarration track generated: {output_filename}")

produce_audio_narration(marketing_text)
```
