```python
# Final script to run everything together for the pet assistant project
import torch
from transformers import pipeline
from diffusers import StableDiffusionPipeline
from gtts import gTTS
from IPython.display import display

# Setting up my specific project product focus
my_pet_assistant = "AI-powered smart pet assistant that monitors pet health and behavior"

# A. Generate the ad text for social media
text_gen = pipeline("text-generation", model="gpt2")
post_draft = text_gen(f"Write an engaging ad for my {my_pet_assistant}:", max_length=80, num_return_sequences=1)
ad_text = post_draft[0]['generated_text']

# B. Generate the image using CPU-friendly settings
# I'm using torch.float32 here so it runs correctly on my AMD setup
pipe = StableDiffusionPipeline.from_pretrained("runwayml/stable-diffusion-v1-5", torch_dtype=torch.float32).to("cpu")
pet_image = pipe("A sleek, modern smart device sitting next to a cute healthy dog, professional lighting").images[0]
pet_image.save("pet_assistant_promo.png")

# C. Generate the audio narration
tts = gTTS(text=ad_text, lang='en', slow=False)
tts.save("pet_assistant_audio.mp3")

# Display the results to verify the pipeline worked
print("Ad Copy generated:", ad_text)
display(pet_image)
print("Files saved: pet_assistant_promo.png and pet_assistant_audio.mp3")
```
